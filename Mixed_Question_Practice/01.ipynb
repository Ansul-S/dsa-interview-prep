## 🧠 Question 1 — Arrays (Medium)

### 🔹 Rearrange Positives and Negatives

Given an array of positive and negative integers, move all negative numbers to the left and all positive numbers to the right.  
**Order within each group doesn't matter.**

---

### 📥 Input
[3, -1, 2, -4, 5, -2]


### 📤 Output
[-1, -4, -2, 3, 2, 5] (any valid arrangement)


---

### ⚙️ Constraints
- ⏱️ Time Complexity: **O(n)**
- 💾 Space Complexity: **O(1)**
- 🔁 Must be **in-place**
- 🚫 No built-in sort allowed

---

### 💡 Hint (Hover to Reveal)

<details>
<summary>Click to reveal hint</summary>

Think about which **Module 1 pattern** moves elements to one side based on a condition.  
You've done this before with **zeroes**.

</details>


```python
arr = [3,-1,2,-4,5,-2]
def Rearrange(arr):
    slow = 0

    for fast in range(len(arr)):
        if arr[fast] < 0:
            arr[slow], arr[fast] = arr[fast], arr[slow]
            slow = slow + 1
    return arr
print(Rearrange(arr)) 
```

### Problem
Rearrange negatives left, positives right

### Pattern
Two pointers — slow/fast (same direction)

### Key Trick
Exact same as *Move Zeroes*  
Just change the condition:
- `!= 0` → `< 0`

### Mistake I made
None ✅

### When to use
Any partition problem —  
separate elements by any condition
