## 🧠 Question 1 — Arrays (Medium)

### 🔹 Rearrange Positives and Negatives

Given an array of positive and negative integers, move all negative numbers to the left and all positive numbers to the right.  
**Order within each group doesn't matter.**

---

### 📥 Input
[3, -1, 2, -4, 5, -2]


### 📤 Output
[-1, -4, -2, 3, 2, 5] (any valid arrangement)


---

### ⚙️ Constraints
- ⏱️ Time Complexity: **O(n)**
- 💾 Space Complexity: **O(1)**
- 🔁 Must be **in-place**
- 🚫 No built-in sort allowed

---

### 💡 Hint (Hover to Reveal)

<details>
<summary>Click to reveal hint</summary>

Think about which **Module 1 pattern** moves elements to one side based on a condition.  
You've done this before with **zeroes**.

</details>


```python
arr = [3,-1,2,-4,5,-2]
def Rearrange(arr):
    slow = 0

    for fast in range(len(arr)):
        if arr[fast] < 0:
            arr[slow], arr[fast] = arr[fast], arr[slow]
            slow = slow + 1
    return arr
print(Rearrange(arr)) 
```

### Problem
Rearrange negatives left, positives right

### Pattern
Two pointers — slow/fast (same direction)

### Key Trick
Exact same as *Move Zeroes*  
Just change the condition:
- `!= 0` → `< 0`

### Mistake I made
None ✅

### When to use
Any partition problem —  
separate elements by any condition

## 🧠 Question 2 — Singly Linked List (Medium)

### 🔹 Find the Nth Node From the End

Given a linked list and a number `n`, return the value of the **nth node from the end** — in **one pass only**.

---

### 📥 Input
1 → 2 → 3 → 4 → 5, n = 2


### 📤 Output
4 (2nd from end)


---

### 📥 Input
1 → 2 → 3 → 4 → 5, n = 1

### 📤 Output
5 (last node)


---

### ⚙️ Constraints
- ⏱️ Time Complexity: **O(n)**
- 💾 Space Complexity: **O(1)**
- 🔁 Only **one pass allowed** (no pre-counting length)

---

### 💡 Hint (Hover to Reveal)

<details>
<summary>Click to reveal hint</summary>

You have **two pointers** available.  
What if one pointer got a **head start of `n` steps**?

</details>

```python
class Node():
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList():
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        current = self.head
        while current.next != None:
            current = current.next
        current.next = new_node

    def nodeFromEnd(self, n):
        fast = self.head
        slow = self.head

        for _ in range(n):
            fast = fast.next
        
        while fast:
            fast = fast.next
            slow = slow.next

        return slow.data

ll = LinkedList()
for i in range(1, 6):
    ll.append(i)

print(ll.nodeFromEnd(1)) 
print(ll.nodeFromEnd(2)) 
print(ll.nodeFromEnd(5)) 
```

### Problem: Find Nth node from end

- **Pattern:** Two pointers — gap of `n` steps  
- **Key Trick:**  
  - Move fast `n` steps first  
  - Then move both together  
  - When `fast == None` → `slow` is answer  

- **Mistake I made:** None ✅  

- **When to use:**  
  - Any "from the end" problem  
  - Without knowing list length  

- **Time:** O(n)  
- **Space:** O(1)

## 🧠 Question 3 — Linked List (Hard / Interview Level)

### 🔹 Check if Linked List is a Palindrome

Given a linked list, return `True` if it reads the same forwards and backwards.

---

### 📥 Input
1 → 2 → 3 → 2 → 1


### 📤 Output
True


---

### 📥 Input
1 → 2 → 3 → 4 → 5


### 📤 Output
False


---

### 📥 Input
1 → 2 → 2 → 1



### 📤 Output
True


---

### ⚙️ Constraints
- ⏱️ Time Complexity: **O(n)**
- 💾 Space Complexity: **O(1)**

---

### 💡 Hint (Hover to Reveal)

<details>
<summary>Click to reveal hint</summary>

You've already solved two problems that together crack this one:

1. **Find the middle of the linked list** (fast & slow pointers)
2. **Reverse the second half**

Now compare both halves.

</details>


```python
class Node():
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList():
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        current = self.head
        while current.next != None:
            current = current.next
        current.next = new_node

    def is_palindrome(self):
        if self.head is None or self.head.next is None:
            return True
        
        slow = self.head
        fast = self.head

        # Step 1: find middle
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        # Step 2: Reverse second half
        prev = None
        current = slow
        while current:
            next_node = current.next
            current.next = prev
            prev = current
            current = next_node

        # Step 3: Compare halves
        first = self.head
        second = prev

        while second: # Only need to check second half
            if first.data != second.data:
                return False
            first = first.next
            second = second.next

        return True
    
def test_palindrome(arr):
    ll = LinkedList()
    for x in arr:
        ll.append(x)

    print(f"Input: {arr}")
    print("Is Palindrome:", ll.is_palindrome())
    print("-" * 30)


# Run tests
test_palindrome([1, 2, 3, 2, 1])   
test_palindrome([1, 2, 2, 1])   
test_palindrome([1, 2, 3])     
test_palindrome([1])               
test_palindrome([])           
```

#### Problem: Check if Linked List is Palindrome

**Pattern:**  
Find middle + Reverse + Compare  

**Key Trick:**  
Compare second half only → handles both odd and even lengths  

---

#### Three Steps:
1. Use slow/fast pointers → find middle  
2. Reverse from middle → `prev` becomes new head  
3. Traverse first half and reversed half together → compare  

---

#### Edge Cases:
- Empty list → True  
- Single node → True  

---

#### Mistake I Made:
None ✅  

---

#### Complexity:
- Time: `O(n)`  
- Space: `O(1)`
